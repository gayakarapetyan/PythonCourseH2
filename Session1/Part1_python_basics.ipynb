## 7. Loops

A loop repeats a block of code multiple times. Instead of writing the same line 7 times for each day of the week, you write it once and let the loop handle the repetition.

There are two types of loops in Python: `for` and `while`.

### 7.1 The for loop with range()

`range()` generates a sequence of numbers. It is the simplest way to repeat something a fixed number of times.

```python
range(5)        # 0, 1, 2, 3, 4
range(1, 6)     # 1, 2, 3, 4, 5
range(0, 10, 2) # 0, 2, 4, 6, 8  (step of 2)
range(5, 0, -1) # 5, 4, 3, 2, 1  (counting down)
```

````python
# Basic range loop
for i in range(5):
    print(i)
````

````python
# Using the index to access list elements
temperatures = [13.5, 14.2, 12.8, 8.7, 10.2, 12.5, 13.9]

for i in range(len(temperatures)):
    print(f"Day {i}: {temperatures[i]} C")
````

### 7.2 The for loop over a list directly

When you do not need the index number, you can loop directly over the values. This is cleaner and more readable.

````python
# Looping directly over values
temperatures = [13.5, 14.2, 12.8, 8.7, 10.2, 12.5, 13.9]

for temp in temperatures:
    print(f"{temp} C")
````

````python
# Combining a loop with a condition
# Count how many days were below 10 C
cold_days = 0

for temp in temperatures:
    if temp < 10:
        cold_days = cold_days + 1

print(f"Cold days this week: {cold_days}")
````

### 7.3 The while loop

A `while` loop keeps running as long as a condition is true. You use it when you do not know in advance how many times to repeat.

Be careful — if the condition never becomes false, the loop runs forever.

````python
# A while loop counts up until a condition is met
reading = 50   # starting electric field value

while reading < 400:
    print(f"E-field: {reading} V/m — still calm weather")
    reading = reading + 100

print(f"E-field: {reading} V/m — threshold reached")
````

### 7.4 List comprehension

List comprehension is a compact way to build a new list from an existing one. It is one line instead of three or four.

The pattern is:
```python
new_list = [expression for item in old_list]
```

You can also add a condition:
```python
new_list = [expression for item in old_list if condition]
```

````python
# Convert all temperatures from Celsius to Fahrenheit
temps_c = [13.5, 14.2, 12.8, 8.7, 10.2, 12.5, 13.9]

# Without list comprehension:
temps_f = []
for t in temps_c:
    temps_f.append(t * 9/5 + 32)
print(temps_f)

# With list comprehension — same result, one line:
temps_f = [t * 9/5 + 32 for t in temps_c]
print(temps_f)
````

````python
# Keep only the values above 13 C
warm_days = [t for t in temps_c if t > 13]
print(warm_days)
````

### 7.5 zip() — looping over two lists at once

So far we have been looping over one list at a time. But often you have two parallel lists — days and temperatures, for example — and you want to use both together.

`zip()` pairs up the elements from two lists so you can loop through them simultaneously.

````python
# First the range() approach — works but a bit clunky
days  = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temps = [13.5,  14.2,  12.8,  8.7,  10.2,  12.5,  13.9]

for i in range(len(days)):
    print(f"{days[i]}: {temps[i]} C")
````

## 8. Functions

Look at the classification code we just wrote. What if we need to classify an electric field value in 10 different places in our program? We would have to write those three `if/elif/else` lines 10 times.

A function solves this. You write the logic once, give it a name, and call it whenever you need it.

The structure is:
```python
def function_name(parameter):
    # do something
    return result
```

- `def` tells Python you are defining a function
- the **parameter** is the input value the function receives
- `return` sends the result back to wherever the function was called

````python
# Our electric field classifier as a function
def classify_efield(value):
    """Classify atmospheric electric field in V/m."""
    if value > 400:
        return "storm"
    elif value >= 100:
        return "disturbed"
    else:
        return "fair weather"

# Now we can call it anywhere
print(classify_efield(650))
print(classify_efield(210))
print(classify_efield(85))
````

````python
# Using the function inside a loop — the loop becomes much cleaner
days    = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
efields = [85, 210, 95, 650, 430, 110, 78]

for day, ef in zip(days, efields):
    print(f"{day}: {ef} V/m -> {classify_efield(ef)}")
````

````python
# Functions can have multiple parameters
def describe_day(day_name, temp, rain):
    """Print a summary of one day."""
    print(f"{day_name}: {temp:.1f} C, {rain:.1f} mm rain")

describe_day("Thursday", 8.7, 45.3)
describe_day("Friday",  10.2, 12.7)
````

````python
# Default parameters
# You can give a parameter a default value so the caller does not always have to provide it

def classify_efield(value, storm_threshold=400, disturbed_threshold=100):
    """Classify atmospheric electric field in V/m."""
    if value > storm_threshold:
        return "storm"
    elif value >= disturbed_threshold:
        return "disturbed"
    else:
        return "fair weather"

# Using defaults
print(classify_efield(650))

# Overriding the storm threshold
print(classify_efield(650, storm_threshold=800))
````

### 8.1 Functions with calculations

Functions are not just for classification. You can put any calculation inside one and return the result. This is very useful for physical formulas you want to use repeatedly.

````python
# Dew point approximation using the Magnus formula
# The dew point is the temperature at which air becomes saturated
# This is a simplified version: dew_point = temp - ((100 - humidity) / 5)

def dew_point(temp_c, humidity_pct):
    """Estimate dew point temperature in Celsius."""
    return temp_c - ((100 - humidity_pct) / 5)

dp = dew_point(13.5, 78)
print(f"Dew point: {dp:.1f} C")

# If temperature is close to dew point, fog or condensation is likely
temp = 8.7
hum  = 94
dp   = dew_point(temp, hum)
print(f"Temp: {temp} C  |  Dew point: {dp:.1f} C  |  Difference: {temp - dp:.1f} C")
````

## 10. Classes

Before we talk about classes, think about what you already know:

- A **dictionary** stores data together under one name
- A **function** does something with data

A **class** is simply those two things combined. Data and the functions that work on that data, bundled together into one object.

To understand why this is useful, we start with a simple example that has nothing to do with weather.

### 10.1 Starting simple — the Apple example

Imagine you are working with apples. Each apple has a color and a weight. You want to be able to check if an apple is heavy.

````python
# Without a class — data and functions are separate
apple1 = {"color": "red",   "weight_g": 182}
apple2 = {"color": "green", "weight_g": 230}

def is_heavy(apple):
    return apple["weight_g"] > 200

print(is_heavy(apple1))   # False
print(is_heavy(apple2))   # True

# This works, but as your program grows,
# you end up with dozens of dictionaries and dozens of functions
# all floating around separately. Hard to manage.
````

````python
# With a class — data and functions live together
class Apple:

    def __init__(self, color, weight_g):
        self.color    = color
        self.weight_g = weight_g

    def is_heavy(self):
        return self.weight_g > 200


apple1 = Apple("red",   182)
apple2 = Apple("green", 230)

print(apple1.is_heavy())   # False
print(apple2.is_heavy())   # True
print(apple1.color)        # red
print(apple2.weight_g)     # 230
````

Two things to understand here:

**`__init__`** is the function that runs automatically when you create a new object. It sets up the initial data.

**`self`** means "this particular apple". When you have 10 apples and call `apple2.is_heavy()`, Python needs to know which apple's weight to check. `self` is how the method knows which object it belongs to. You always write `self` as the first parameter of every method, but you never pass it yourself — Python handles it automatically.

### 10.2 Building up a WeatherStation class step by step

Now we do the same thing with something meaningful. We will build the class gradually — start with just the data, then add methods one at a time.

````python
# Step 1 — just the data
class WeatherStation:

    def __init__(self, name, latitude, longitude):
        self.name      = name
        self.latitude  = latitude
        self.longitude = longitude


station = WeatherStation("Magdeburg", 52.13, 11.62)
print(station.name)
print(station.latitude)
````

````python
# Step 2 — add a list to store readings, and a method to add them
class WeatherStation:

    def __init__(self, name, latitude, longitude):
        self.name      = name
        self.latitude  = latitude
        self.longitude = longitude
        self.readings  = []   # starts empty

    def add_reading(self, temp_c, rain_mm):
        self.readings.append({"temp_c": temp_c, "rain_mm": rain_mm})

station = WeatherStation("Magdeburg", 52.13, 11.62)
station.add_reading(13.5, 0.0)
station.add_reading(8.7,  45.3)
station.add_reading(10.2, 12.7)

print(f"Number of readings: {len(station.readings)}")
print(station.readings)
````

````python
# Step 3 — add methods that calculate things from the readings
class WeatherStation:

    def __init__(self, name, latitude, longitude):
        self.name      = name
        self.latitude  = latitude
        self.longitude = longitude
        self.readings  = []

    def add_reading(self, temp_c, rain_mm):
        self.readings.append({"temp_c": temp_c, "rain_mm": rain_mm})

    def mean_temperature(self):
        temps = [r["temp_c"] for r in self.readings]
        return sum(temps) / len(temps)

    def total_rain(self):
        return sum(r["rain_mm"] for r in self.readings)

    def summary(self):
        print(f"Station:       {self.name}")
        print(f"Readings:      {len(self.readings)}")
        print(f"Mean temp:     {self.mean_temperature():.1f} C")
        print(f"Total rain:    {self.total_rain():.1f} mm")


station = WeatherStation("Magdeburg", 52.13, 11.62)
station.add_reading(13.5, 0.0)
station.add_reading(14.2, 2.1)
station.add_reading(12.8, 0.0)
station.add_reading(8.7,  45.3)
station.add_reading(10.2, 12.7)
station.add_reading(12.5, 0.0)
station.add_reading(13.9, 3.4)

station.summary()
````

Write code that:
1. Defines a function `week_stats(data)` that returns the mean, min, and max of a list
2. Calls it for each week and prints a formatted report
3. Finds and prints which week was the warmest
4. Counts how many days across all three weeks were below 5 C

There is no single correct solution. Focus on writing clean, readable code.

````python
# Your solution here
week1 = [13.5, 14.2, 12.8, 8.7, 10.2, 12.5, 13.9]
week2 = [9.1,  8.3,  10.5, 11.2, 9.8,  8.7,  10.1]
week3 = [4.2,  3.8,  5.1,  6.3,  4.9,  5.5,  6.1]
````

````python
# ================================================
# SOLUTION — Final Task
# ================================================
week1 = [13.5, 14.2, 12.8, 8.7, 10.2, 12.5, 13.9]
week2 = [9.1,  8.3,  10.5, 11.2, 9.8,  8.7,  10.1]
week3 = [4.2,  3.8,  5.1,  6.3,  4.9,  5.5,  6.1]

def week_stats(data):
    """Return mean, min and max of a list."""
    mean = sum(data) / len(data)
    return mean, min(data), max(data)

weeks = [week1, week2, week3]
means = []

print("Weekly temperature summary")
print("-" * 40)

for i, week in enumerate(weeks):
    mean, low, high = week_stats(week)
    means.append(mean)
    print(f"Week {i+1}: mean={mean:.1f} C  min={low} C  max={high} C")

print("-" * 40)

warmest = means.index(max(means)) + 1
print(f"Warmest week: Week {warmest}")

all_temps  = week1 + week2 + week3
cold_count = len([t for t in all_temps if t < 5])
print(f"Days below 5 C across all weeks: {cold_count}")
````
